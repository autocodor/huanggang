```python
def addsub_value_thread(self, value, target, devicename, add_name, sub_name, tem, key, delay=5):
        """线程函数：逐步修改设备值（支持浮点数）"""
        try:
            current = round(value)
            target = round(target)

            self.logger.debug(f"开始修改: {current}->{target}")

            for i in range(abs(current-target)):
                if target > current:
                    data = {devicename: {add_name: 1}}
                else:
                    data = {devicename: {sub_name: 1}}

                if tem < -10 and tem > 45:
                    self.logger.debug('风机温度异常，跳出循环')
                    break

                self.client.publish(self.m_cfng['write_topic'], data)
                self.logger.debug(f"修改成功: {data}，实时反馈：{self.m_cfng[f'fj{key}_fk']}")

                time.sleep(delay)  # 确保这行与循环内的其他代码缩进一致

            self.logger.debug(f"修改完成: {target}")
        except Exception as e:
            self.logger.error(f"修改值错误: {str(e)}")
```
